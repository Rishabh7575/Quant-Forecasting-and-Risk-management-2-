# Customer Churn Analysis EDA
This notebook contains an exploratory data analysis (EDA) of the IBM Telco Customer Churn dataset.
We will visualize various factors that may influence a customer's decision to churn.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset
df = pd.read_csv('../data/processed/churn_clean.csv')
df.head()
```

## 1. Churn Distribution
Let's look at the overall distribution of customer churn.

```
plt.figure(figsize=(8, 6))
sns.countplot(x='Churn', data=df, palette='Set2')
plt.title('Churn Distribution')
plt.show()
```

**Observation:**
The dataset is imbalanced. The number of customers who did not churn is significantly higher than those who did churn.

## 2. Gender vs Churn
Does gender have an impact on the likelihood of a customer churning?

```
plt.figure(figsize=(8, 6))
sns.countplot(x='gender', hue='Churn', data=df, palette='Set2')
plt.title('Gender vs Churn')
plt.show()
```

**Observation:**
The churn rate is roughly the same for both males and females, indicating that gender may not be a strong predictor of churn.

## 3. Contract Type vs Churn
How does the customer's contract type affect their churn status?

```
plt.figure(figsize=(8, 6))
sns.countplot(x='Contract', hue='Churn', data=df, palette='Set2')
plt.title('Contract Type vs Churn')
plt.show()
```

**Observation:**
Customers with a 'Month-to-month' contract have a much higher churn rate compared to those with 'One year' or 'Two year' contracts. This is a very strong signal.

## 4. Monthly Charges Distribution
Let's see the distribution of monthly charges among the customers.

```
plt.figure(figsize=(8, 6))
sns.histplot(df['MonthlyCharges'], kde=True, bins=30, color='purple')
plt.title('Monthly Charges Distribution')
plt.show()
```

**Observation:**
A large number of customers have lower monthly charges (around $20). There is also a wide spread up to $120. Higher charges might be correlated with higher churn, which warrants further investigation.

## 5. Tenure Distribution
How long do customers typically stay?

```
plt.figure(figsize=(8, 6))
sns.histplot(df['tenure'], kde=True, bins=30, color='teal')
plt.title('Tenure Distribution')
plt.show()
```

**Observation:**
Many customers have a very low tenure (just 1 or 2 months), representing new customers or rapid churners. Another spike exists at around 70+ months, representing loyal, long-term customers.
